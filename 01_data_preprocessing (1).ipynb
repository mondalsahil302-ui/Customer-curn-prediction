```python
df = pd.read_csv("IBM-Telco-Customer-Churn.csv")

print("Dataset loaded successfully")
print(df.shape)
```

```python
df_clean = df.copy()
```

```python
df_clean.info()
```

```python
(df_clean == " ").sum()
```

```python
df_clean["TotalCharges"] = (
    df_clean["TotalCharges"]
    .astype(str)
    .str.strip()
)

df_clean["TotalCharges"] = df_clean["TotalCharges"].replace("", np.nan)
```

```python
df_clean["TotalCharges"] = pd.to_numeric(
    df_clean["TotalCharges"],
    errors="coerce"
)

print(df_clean["TotalCharges"].dtype)
```

```python
df_clean = df.dtypes
df_clean
```

```python
df_clean.isnull().sum()
```

```python
median_value = df_clean["TotalCharges"].median()

df_clean["TotalCharges"] = df_clean["TotalCharges"].fillna(median_value)

df_clean.isnull().sum()
```

```python
median_value = df_clean["TotalCharges"].median()

df_clean["TotalCharges"] = df_clean["TotalCharges"].fillna(median_value)

df_clean.isnull().sum()
```

```python
print("Duplicate rows:", df_clean.duplicated().sum())

df_clean = df_clean.drop_duplicates()

print("Shape after removing duplicates:", df_clean.shape)
```

```python
for col in df_clean.select_dtypes(include=["object"]).columns:
    df_clean[col] = df_clean[col].str.strip()

print("Text columns cleaned")
```

```python
df_clean["TenureGroup"] = pd.cut(
    df_clean["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=[
        "0-1 Year",
        "1-2 Years",
        "2-4 Years",
        "4-6 Years"
    ]
)

df_clean[["tenure", "TenureGroup"]].head()
```
